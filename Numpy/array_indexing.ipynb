```python
import numpy as np
```

 # Numpy Indexing and Slicing of vectors

```python
arr = np.arange(11,21)
arr
```

```python
# indexing
print(arr[6])
```

```python
# slicing
# arr[start:endpoint:]
arr[1:5]
arr[:5] # default start value
arr[:] # auatmatic at last 
arr[3:]
arr_slice = arr[3::2] # arr[start::steps]
arr_slice
```

# Numpy Indexing and Slicing of Vectors

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5] # rows target -> 0,1,2,3,4,5
```

```python
print(arr[0,0])
```

```python
print(arr[0,4])
```

```python
print(arr[5,4])
```

```python
# slicing
# 2,3
# 7,8
slice = arr[0:2,1:3] # arr[row,col]
slice
```

```python
arr[3:6,3:6] # or
arr[3:,3:]
```

```python
arr[:,2 # created vector
```

# Boolean Indexing

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr% 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
