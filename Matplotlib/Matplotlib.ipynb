MATLAB is a high-level programming language and a numerical computing environment used by engineers and scientists 
for data analysis, visualisation, and algorithmic development. It's known for its intuitive syntax, particularly for working with matrices and arrays, making it a popular choice for mathematical and scientific computing.

```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 
```

## Basics

```python
There are 10 gaps:
0 ── 0.5 ── 1 ── 1.5 ── 2 ── 2.5 ── 3 ── 3.5 ── 4 ── 4.5 ── 5

The gap is:

(5 - 0) / (11 - 1)
= 5 / 10
= 0.5

So every number increases by 0.5.

arange

np.arange(0, 5, 0.5)

means "start at 0, increase by 0.5, and stop before 5."

So linspace focuses on how many values you want, while arange focuses on the step size.
```

```python
x = np.linspace(0,5,11) 
x
y = x **2
y
```

```python
plt.plot(x,y)
```

```python
# adding levels ,legends
```

```python
plt.title('hii')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

# object oriented way


Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

```python
# fig = plt.figure() # selcet convas
# fig.add_axes([0.1,0.1,0.8,0.8]) #[left,right,width,heigth]
```

```python
fig = plt.figure(figsize=(20,8),dpi = 100) # depth per inch
# fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)

# axis2 = fig.add_axes([0.1,0.1,.8,.8])
axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
axis2.plot(x,y)
```
