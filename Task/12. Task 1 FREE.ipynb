### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2


````python
# Write your code here
#  formula for this is => 
# heads = 4, legs = 12
# dogs + chickens = 4
# 4 * dogs + 2* chickens = 12
x = int(input("Enter the value of heads:"))
y = int(input("enter the legs: "))
dogs = int((y-2*x)/2)
chickens = int((x-dogs))
print(dogs,chickens)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.


````python
# Write your code here
# Formula for sum of squares of natural number is =>
# n(n+1)(2n+1)/6
n = int(input("Enter the value of n:"))
sum = float((n*(n+1)*(2*n+1)/6))
print(sum)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.


````python
# Write your code here
# fomula fro nth term is a +(n-1) * d, d = b-a
n = int(input("Enter the value of n:"))
a = int(input("Enter the value of a:"))
b = int(input("Enter the value of b:"))
d = int(b-a)
NthTerm = a + (n-1) * d
print(NthTerm)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.

Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm


````python
# Write your code here
# Tank is cuboid , Volume = h * l * b
# Glass is a cylinder,Volume = pi * r^2 * h
# glass = Tank Volume / Glass Volume
import math
H_t = int(input("Enter the value of Height"))
L_t = int(input("Enter the value of length"))
B_t = int(input("Enter the value of Bridth"))
r_g = float(input("Enter the radius of glass: "))
h_g = float(input("Enter the height of glass: "))
Vol_t =  H_t * L_t * B_t
Vol_g = 3.14 * r_g**2 * h_g
print('no of glasses',math.floor(Vol_t/Vol_g))
````
