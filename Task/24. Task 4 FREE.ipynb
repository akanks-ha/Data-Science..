###`Problem 1:` Combine two lists index-wise(columns wise)

`Given List:`

```
list1 = ["M", "na", "i", "Kh"]

```

`Output:`

```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`


````python
# Write code here
l = [2,4,6,10,1]


res = []

for i in l : # i = 2 , 4 ,5,10,1
    sum = 0
    for j in l:  #  j = 2,4,5,10,1 => i = 2 => j = (2,4,5,10,1), i = 4 => j = ( 2,4,5,10,1) etc.
        if i <= j:
            sum += j
    res.append(sum)

print(res)
                                                                                                                                                                                                                                                                                                                                                                                                                                              
````

###`Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`

```
['1ac21', '23fg', '456', '098d','1','kls']
```

`Output:`

```
['456', '23fg', '1ac21', '1', 'kls', '098d']

```


````python
# Write code here
l = [ '1ac21','23fg','456','098d','1','kls']
res = []
for item in l:
    prod = 1
    for char in item:
        if char.isdigit():
            prod = prod*int(char)
    res.append(prod)
# sorted(list(zip(res,l)),reverse = True) => kronki hme bas strinf chiye use list comprehension

[i[1] for i in sorted(list(zip(res,l)),reverse = True)]
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:

```bash
['campusx Is', 'best For', 'data Scientist']
```


````python
# Code here
# Potential Words => means you have to find when you see cap letter words
l = ['capusxIs','bestFor','dataScientist']

res= []

for i in l :
    temp = [[]]
    for char in i:
        
        if char.isupper():
            temp.append([])
        
        temp[-1].append(char)
    
    temp_str = ''
    for item in temp:
        temp_str = temp_str + "".join(item) + " "
    res.append(temp_str[0:-1])

print(res)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:

```bash
[1,2,3,4,5,7,8]
```


````python
# Write code here
l1 = [1,2,3,4,5,1]
l2 = [2,3,5,7,8]

union = []

for i in l1 :
    if i not in union:
        union.append(i)
for j in l2:
    if j not in union:
        union.append(j)
print(union)
````
