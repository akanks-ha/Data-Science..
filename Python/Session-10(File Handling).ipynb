* If we dont it.garbage collector would does it
* with keyword closes the file as soon as the usage is over

# Serilization andDeserialization
* Serialization - process of coverting python data types to JSON formate
* Deserialization - process of converting JSON to python data types

What is JSON? (java  script on notation)

```python
# serilization using json module
# list
import json
L = [1,2,3,4]

with open('demo.json','w') as f:
  json.dump(L,f)
```

```python
# dict
d = {
    'name' : 'aku',
    'age' : 23,
    'gender' : 'female'
    }
with open('demo.json','w') as f:
  # json.dump(d,f)
  # for better
  json.dump(d,f,indent = 4)
```

```python
# deserialization
import json

with open('demo.json','r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

```python
# serialization and deserialize tuple
import json

t = (1,2,3,4,5)

with open('demo.json','w') as f:
  json.dump(t,f)
```

# Serializating and Deserizing custom objects


```python
class Person:

  def __init__(self,fname,lname,age,gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

# format to printes in
# -> Aku age -> 23 gender -> female
```

```python
person = Person('aku','UP',23,'female')
```

```python
# Aa a string
import json

def show_object(person):
  if isinstance(person,Person):
    return{'name':person.fname + " "+person.lname,'age':person.age,'gender':person.gender}

with open('demo.json','w') as f:
  json.dump(person,f,default = show_object,indent = 4)
```

```python
# indent attribute
# As a dict
```

```python
# deserializing
import json

with open('demo.json','r') as f:
  d=json.load(f)
  print(type(d))
```

## Pickling

Pickling is the process whereby a Python object hierarchy is converted into a byte stream,and unpickling is the inverse operation , whereby a byte stream(from a binary file of byte-like object) is converted back into an object hierarchy.

```python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am',self.age,'years old')
```

```python
p = Person('aku',23)
```

```python
# pickle dump
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
  p = pickle.load(f)

p.display_info()
```
