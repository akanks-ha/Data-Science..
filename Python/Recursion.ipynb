```python
def multiply(a,b):
  result = 0

  for i in range(b):
    result = result + a
  print(result)

multiply(3,4)

def multi(a,b):
  return a* b

print(multi(3,4))
```

```python
def mul(a,b):

  if b==1:
    return a
  else:
    return a+ mul(a,b-1)

print(mul(5,6))
```

```python
def fact(num):
  fact = 1
  if num == 0 :
    return 1
  else:
    return fact * (fact)

print(fact(5))
```

```python
# palindrome
def palindrome(text):
  if len(text) <=1:
    print('palindrone')
  else:
    if text[0] == text[-1]:
      palindrome(text[1:-1])
    else:
      print("Not a palindrome")

palindrome("madam")
palindrome("malayalam")
palindrome("python")
palindrome("abba")

```

```python
# rabbit problem
# If 2 newborn rabbits are put in a pen,how many rabbits will be in the pen after 1 year?
# Asuume that rabbits always produce one male and female offspring .... can reproduce in every month ... can reproduce once they are one month old... never die!

# this is the problem on the pattern of fibonacci

def fibo(m):
  if m==0 or m==1:
    return 1
  else:
    return fibo(m-1) + fibo(m-2)
print(fibo(12))
```

```python
# using above fibonacci code when i'm increasing no of m then system get slow
# memoization -> DP Problem
import time
def memo(m,d):

  if m in d:
    return d[m]
  else:
    d[m] = memo(m-1,d) + memo(m-2,d)
    return d[m]
start = time.time()
d = {0:1,1:1}
print(memo(500,d))
print(time.time()-start)
print(d)
```
