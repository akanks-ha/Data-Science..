# dir/instance/isdubclass

```python
# dir

class Test:
  def __init__(self):
    self.foo = 11
    self.bar = 23
    self.__baz = 23


  def greet(self):
    print('hello')

t = Test()
print(dir(t)) # This gives us a list with the objcet's attributes
```

```python
# isinstance

class Example:

  def __init__(self):
    print('hello')

obj = Example()

isinstance(obj,Example)
```

```python
# issubclass
class A:
  def __init__(self):
    pass

class B(A):
  pass

issubclass(A,B)
```

# The diamond Problem

```python
class Class1:
  def m(self):
    print("In class1")

class Class2(Class1):
  def m(self):
    print("In Class2")

class Class3(Class1):
  def m(self):
    print("In Class3")

class Class4(Class2,Class3): # yha pe amibiquity name ki cheej nhi hoti hai jo bhi order rhega uske according work krega callef #MRO(Method Resolution Order)
  pass

obj = Class4()
obj.m()
```

# What's the meaning of single and double underscores in Python variable and method names

* Single Leading Underscore:_var
* Single Trailing Underscore:var_
* Double Leading Undersocre:__Var
* Double LEading and Trailing Underscore:__var__
* Single Undersocre:_
1. Single Leading Underscore:_var are a Python naming convention that  indicates a name is meant for internal use. It is geenarlly not enforced by the Python interpreter and is only meant as a hint to the programmer.

Adding a single underscore in front of a variable name is more like someone putting up a tiny underscore warnning sign that says:

"Hey, this isn't really meant to ba part of the public interface of this class.Best to leave it alone."

```python
class A:

  def __init__(self):
    self._var = 10

a = A()
a._var
```

```python
class Test:
  def __init__(self):
    self.foo = 11
    self._bar = 23

t = Test()
print(t.foo) # print 11

print(t._bar) # print 23


```

2. Single Trailing Underscore:var_ Sometimes the most fiting name for a variable is already taken by a keyword in the Python language.Therefor,names like class or def cannot be used as variable names in Python. In this case ,you can append a single underscoew to break the naming conflict:

```python
def make_objet(name,class):
  pass
```

In summary,a single trailing underscore (postfix) is used by convention is defined and explained in PEP 8.



3. Double Leading Undersocre:__var

A doubleunderscore prefix causes the Python interpreter to rewrite the attributr name in oreder to avoid naming conflicts in subclasses.

This is also called name managling-the interprete changes the name of the varible in a way that makes it harder to create collions when the class is extended later.

4. Double Leading and Trailing Undersocre:__var__

Double underscore __ are often refered to as 'dunders' in the Python community.The reason is that double unersocres appear quite often in Python code,and to avoid fatigue their jaw muscles,Pythonistas often shorten 'double inderscore' to 'dunder'.

The name that have both leading and trailing double underscore are reserved for special use in the language.This rule covers things like __init__ for object constructors,or __cal__ to make objects callable.

These dunder methods are often refered to as magic methods.

There are many dunder methods,here are some:
__str__,__repr__,__call__,__add__,__sub__,__len__ etc.

5. Single Underscore_: Sometimes used as a name for temporary or insignificant variables("don't care).Also,it reoresents the result of the last expression in a Pyhon REPL session.
