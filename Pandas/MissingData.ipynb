```python
import numpy as np 
import pandas as pd 
```

# Finding Missing Data

```python
data = {
    'A' : [1,2,np.nan,4,5],
    'B' : [1,2,3,4,5],
    'C' : [1,2,3,np.nan,np.nan], 
    'D' : [1,np.nan,np.nan,np.nan,5]
}
df = pd.DataFrame(data)
```

```python
# nan = not a number
```

```python
df
```

```python
# find no of nan
df.isna().sum()
```

```python
# any mising value in each columna
df.isna().any() 
```

# Removing Missing Data

works on row basis

```python
df
```

```python
df.dropna()# all rows deleted which have null values
```

```python
df
```

```python
# threshold 
df.dropna(thresh=3) # jiske pas atleast 3 non null values hai unhe chorke sabko dlt kr do
```

# Filling the Missinf Data

```python
df
```

```python
df.fillna(0)
```

```python
df
```

```python
values = {"A" : 0,"B" :100,"C":300,"D":400}
df.fillna(value = values) # columns wise
```
